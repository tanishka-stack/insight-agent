```python
from datasets import load_dataset

dataset = load_dataset("fancyzhx/amazon_polarity")
print(dataset)
```

```python
print(dataset["train"][0])
print(dataset["train"][1])
print(dataset["train"][2])
```

```python
import random

sample_indices = random.sample(range(len(dataset["train"])), 10)
for i in sample_indices:
    ex = dataset["train"][i]
    print(f"Label: {ex['label']} | Title: {ex['title']}")
    print(f"Review: {ex['content'][:300]}")
    print("-" * 80)
```

```python
import pandas as pd

train_df = dataset["train"].to_pandas()
test_df = dataset["test"].to_pandas()

print(train_df.shape)
print(train_df["label"].value_counts())
```

```python
os.makedirs("../data", exist_ok=True)
train_df.to_parquet("../data/amazon_polarity_train.parquet")
test_df.to_parquet("../data/amazon_polarity_test.parquet")
```
